```python
import dash
from dash import dcc, html
from dash.dependencies import Input, Output
import plotly.express as px
import pandas as pd
import requests

# 1. Map Illinois cities to NWS Grid Points (No API key needed!)
CITIES = {
    "Chicago": "https://weather.gov",
    "Springfield": "https://weather.gov",
    "Peoria": "https://weather.gov",
    "Rockford": "https://weather.gov",
    "Champaign": "https://weather.gov"
}

# User-Agent is required by NWS API policy
HEADERS = {"User-Agent": "IllinoisWeatherDashboard/1.0 (your-email@example.com)"}

def fetch_nws_data():
    data = []
    for city, url in CITIES.items():
        try:
            response = requests.get(url, headers=HEADERS, timeout=5)
            if response.status_code == 200:
                forecast_data = response.json()
                # Get the most current hourly forecast period
                current_period = forecast_data['properties']['periods'][0]
                
                # NWS provides relative humidity as a dictionary: {'value': 75, 'unitCode': 'wmoUnit:percent'}
                humidity = current_period.get('relativeHumidity', {}).get('value', 0)
                
                # NWS provides probability of precipitation percentage
                pop = current_period.get('probabilityOfPrecipitation', {}).get('value', 0)
                if pop is None: pop = 0  # Handle null values smoothly
                
                data.append({
                    "City": city,
                    "Humidity (%)": humidity,
                    "Precipitation Probability (%)": pop
                })
            else:
                print(f"NWS Error for {city}: {response.status_code}")
        except Exception as e:
            print(f"Network error for {city}: {e}")
            
    # CRITICAL: Always return a structural template to prevent Plotly "Expected one of []" crash
    if not data:
        return pd.DataFrame([{"City": "No Data", "Humidity (%)": 0, "Precipitation Probability (%)": 0}])
        
    return pd.DataFrame(data)

# 2. Initialize Dash App
app = dash.Dash(__name__)

app.layout = html.Div([
    html.H1("Illinois Weather Dashboard (NWS Live Data)", style={'textAlign': 'center', 'fontFamily': 'sans-serif'}),
    
    html.Div([
        dcc.Graph(id='humidity-bar'),
        dcc.Graph(id='precipitation-bar')
    ], style={'display': 'flex', 'flexDirection': 'column'}),
    
    dcc.Interval(
        id='interval-component',
        interval=300000, # 5 minutes
        n_intervals=0
    )
])

# 3. Callback to Update Graphs
@app.callback(
    [Output('humidity-bar', 'figure'),
     Output('precipitation-bar', 'figure')],
    Input('interval-component', 'n_intervals')
)
def update_graphs(n):
    df = fetch_nws_data()
    
    fig_humidity = px.bar(
        df, x='City', y='Humidity (%)', color='City',
        title="Current Relative Humidity across Illinois Cities"
    )
    
    fig_precip = px.bar(
        df, x='City', y='Precipitation Probability (%)', color='City',
        title="Probability of Precipitation (%)",
        color_discrete_sequence=['#636EFA']
    )
    
    return fig_humidity, fig_precip

if __name__ == '__main__':
    # Setting debug=False prevents the PIN locking prompt if network code behaves erratically
    app.run(debug=False)
```
